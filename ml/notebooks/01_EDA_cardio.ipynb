```python
#feature engineering 


df['bmi'] = df['weight'] / ((df['height'] / 100) ** 2)
df['pulse_pressure'] = df['ap_hi'] - df['ap_lo']
df['map'] = df['ap_lo'] + (1/3) * (df['ap_hi'] - df['ap_lo'])   #MAP -- mean arterial pressure 
```

```python
df = df.drop(["height",'weight','gender'],axis = 1 )
```
